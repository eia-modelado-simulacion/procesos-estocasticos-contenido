## 5. Dos usos diferentes de estacionariedad

**Primero, la intuición:** una distribución de estados puede permanecer igual aunque cada sistema cambie de estado; una serie puede fluctuar aunque sus momentos probabilísticos no cambien con el tiempo.

### En cadenas de Markov

Para la cadena homogénea con matriz $P$, una distribución estacionaria satisface

$$\pi P=\pi,\qquad \pi_i\geq0,\qquad\sum_i\pi_i=1.$$

Si $X_0\sim\pi$, entonces $X_t\sim\pi$ para todo $t\geq0$. Se mantiene la **distribución de estados**, no el estado individual. Esto no exige discutir convergencia desde otra distribución inicial: esa es una pregunta diferente.

### En series temporales

Con segundos momentos finitos, la **estacionariedad débil** requiere, para todos los tiempos y rezagos válidos:

$$E(Y_t)=\mu,$$
$$\operatorname{Var}(Y_t)=\sigma^2<\infty,$$
$$\operatorname{Cov}(Y_t,Y_{t+h})=\gamma(h).$$

La dependencia medida por covarianza depende del rezago $h$, **no del momento particular $t$**. Comparar semanas separadas por dos pasos tiene la misma covarianza teórica al principio y al final del calendario. No basta que solo la media sea constante.

> **La distribución estacionaria de una cadena de Markov y la estacionariedad débil de una serie temporal están relacionadas por la idea de estabilidad probabilística, pero no son conceptos equivalentes.**

Por ejemplo, una cadena homogénea iniciada en $\pi$ produce una serie estacionaria al observar una función numérica fija del estado con segundo momento finito. Pero disponer de $\pi$ no garantiza que cualquier cantidad asociada $Y_t$ sea estacionaria: esa cantidad puede incorporar cambios de nivel o de funcionamiento. Los códigos de categorías tampoco tienen automáticamente distancias numéricas significativas.

## 7. Autocovarianza y autocorrelación

**Intuición.** Imagine que resta el nivel típico a cada observación. Si dos observaciones separadas por $h$ suelen estar ambas por encima o ambas por debajo de ese nivel, sus desviaciones tienen el mismo signo. Si suelen estar a lados opuestos, tienen signos distintos.

Para un proceso débilmente estacionario:

$$\gamma(h)=\operatorname{Cov}(Y_t,Y_{t+h})
=E[(Y_t-\mu)(Y_{t+h}-\mu)].$$

Mide cómo tienden a variar conjuntamente observaciones de la misma serie separadas por $h$ períodos. Sus unidades son las de $Y_t$ al cuadrado. Para comparar sin unidades, si $\gamma(0)=\sigma^2>0$, definimos

$$\rho(h)=\frac{\gamma(h)}{\gamma(0)},\qquad
\gamma(0)=\operatorname{Var}(Y_t),\qquad -1\leq\rho(h)\leq1.$$

En particular, $\rho(0)=1$. Si la varianza es cero, esta normalización no está definida.

| Resultado | Lectura intuitiva en el rezago 1 |
|---|---|
| $\rho(1)>0$ | Persistencia: desviaciones del mismo signo tienden a sucederse. |
| $\rho(1)\approx0$ | Poca asociación **lineal** a ese rezago. |
| $\rho(1)<0$ | Tendencia a alternar a ambos lados de la media; no alternancia obligatoria. |

$$\boxed{\rho(h)\approx0\not\Rightarrow\text{independencia}}$$

La correlación resume asociación lineal. Una relación puede cambiar la magnitud sin favorecer un signo: si $U$ es uniforme en $[-1,1]$ y $V=U^2$, conocer $U$ determina $V$, aunque su covarianza es cero por simetría. Incluso una correlación exactamente cero no asegura independencia.

### De la cantidad teórica a una estimación con datos

No observamos todas las realizaciones para calcular la esperanza. Con una historia estimamos la media y promediamos productos de desviaciones. Usaremos la siguiente convención, con divisor $T$ para todos los rezagos:

$$\bar y=\frac1T\sum_{t=1}^T y_t,\qquad
\widehat\gamma(h)=\frac1T\sum_{t=1}^{T-h}(y_t-\bar y)(y_{t+h}-\bar y),$$

$$\widehat\rho(h)=\frac{\widehat\gamma(h)}{\widehat\gamma(0)}.$$

El sombrero distingue la **estimación muestral** de la cantidad del proceso. Por ejemplo, para $h=2$ usamos los cuatro pares de la actividad anterior. La media se calcula sobre todo el registro; no recalculamos una media diferente para cada extremo del par.

El código reproduce literalmente este cálculo. Exige una serie no constante y $0\leq h<T$. No demuestra estacionariedad: una tendencia o cambio de nivel puede producir una ACF elevada que sería engañosa si se interpretara como dependencia estable.

```python
def acf_muestral(serie, max_rezago):
    """ACF con media global y divisor T en las autocovarianzas."""
    valores = np.asarray(serie, dtype=float)
    T = len(valores)
    if not 0 <= max_rezago < T:
        raise ValueError("El rezago debe estar entre 0 y T - 1.")
    centrada = valores - valores.mean()
    gamma0 = np.sum(centrada ** 2) / T
    if gamma0 == 0:
        raise ValueError("La ACF no está definida para una serie constante.")
    correlaciones = np.empty(max_rezago + 1)
    for h in range(max_rezago + 1):
        gamma_h = np.sum(centrada[:T - h] * centrada[h:]) / T
        correlaciones[h] = gamma_h / gamma0
    return correlaciones

# Ejemplo aritmético de emparejamiento, no diagnóstico de estacionariedad.
print("ACF de la serie corta, h = 0, 1, 2:", acf_muestral(serie_corta, 2))
```

## 8. Función de autocorrelación

Una sola comparación no cuenta toda la historia. Reunamos la correlación de la serie consigo misma a distintas distancias:

$$\boxed{h\mapsto\rho(h)}.$$

Esta es la **función de autocorrelación (ACF)**. La figura calculada con una realización representa $h\mapsto\widehat\rho(h)$, no la función teórica exacta. El eje horizontal muestra rezagos; el vertical, correlaciones sin unidades. El punto $h=0$ vale uno por construcción.

### Actividad 8 — Anticipar el signo sin ver la ACF *(formativa)*

Ejecute **solo la siguiente celda**, que muestra tres series sintéticas. Para A, B y C escriba su anticipación de $\rho(1)$ y señale un tramo que la motive. Discuta con un compañero **antes de avanzar a la celda de ACF**.

A representa ruido independiente. B conserva parte de la desviación anterior y añade ruido. C invierte parte de esa desviación y añade ruido. Compare estas historias para reconocer distintas formas de dependencia; todavía no ajuste modelos para pronosticar.

```python
rng_acf = np.random.default_rng(31415)
T_acf = 240
serie_a = rng_acf.normal(size=T_acf)
serie_b = np.empty(T_acf)
serie_c = np.empty(T_acf)

# Inicialización aleatoria con varianza 1; no hay arranque fijo en cero.
serie_b[0], serie_c[0] = rng_acf.normal(size=2)
peso = 0.8
escala_ruido = np.sqrt(1 - peso ** 2)
for t in range(1, T_acf):
    serie_b[t] = peso * serie_b[t - 1] + escala_ruido * rng_acf.normal()
    serie_c[t] = -peso * serie_c[t - 1] + escala_ruido * rng_acf.normal()

series = [serie_a, serie_b, serie_c]
nombres = ["A — Ruido", "B — Persistencia", "C — Alternancia"]
colores = ["tab:blue", "tab:orange", "tab:green"]
fig, ejes = plt.subplots(3, 1, figsize=(12, 9), sharex=True, sharey=True)
for ax, serie, nombre, color in zip(ejes, series, nombres, colores):
    ax.plot(np.arange(1, T_acf + 1), serie, color=color, linewidth=1)
    ax.axhline(0, color="gray", linewidth=0.8)
    ax.set_title(nombre)
    ax.set_ylabel("Valor observado")
    ax.grid(alpha=0.25)
ejes[-1].set_xlabel("Período t")
plt.tight_layout()
plt.show()
```

<details>
<summary><b>▶ Revelar solución orientadora</b></summary>

A: esperamos una correlación cercana a cero; una muestra de ruido puede exhibir rachas casuales. B: esperamos signo positivo por la continuidad de las desviaciones. C: esperamos signo negativo por la tendencia a invertirlas. No es necesario que cada par siga ese comportamiento.

Las tres simulaciones mantienen media teórica cero y varianza uno. La estabilidad de estos momentos es compatible tanto con ruido independiente como con dependencia positiva o negativa.

</details>

Una barra pequeña en un rezago no elimina la dependencia del mecanismo; la ACF resume únicamente asociación lineal y la estimamos con 100 datos. Sus alturas no son $p_{ij}$: comparan cantidades separadas por semanas, no transiciones entre dos estados particulares.

Además, iniciamos en $X_0=3$, no en una distribución estacionaria. Tener $P$ constante no convierte esta trayectoria en una muestra iniciada en equilibrio; aquí la ACF es una descripción del histórico, no una demostración de estacionariedad ni la ACF teórica de equilibrio.
