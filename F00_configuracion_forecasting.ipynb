## 1. Librerías comunes y configuración

El libro utiliza un bloque de configuración común al inicio de sus capítulos. Aquí se reproduce una versión equivalente para el curso.

```python
import warnings
warnings.filterwarnings("ignore")

import os
os.environ["NIXTLA_ID_AS_COL"] = "true"

import random
import numpy as np
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
import matplotlib as mpl

from cycler import cycler
from utilsforecast.plotting import plot_series as plot_series_utils
from fpppy.utils import plot_series

# Reproducibilidad
np.set_printoptions(suppress=True)
np.random.seed(1)
random.seed(1)

# Configuración de pandas
pd.set_option("max_colwidth", 100)
pd.set_option("display.precision", 3)

# Configuración visual
sns.set_style("whitegrid")
plt.style.use("ggplot")
plt.rcParams.update({
    "figure.figsize": (8, 5),
    "figure.dpi": 100,
    "savefig.dpi": 300,
    "figure.constrained_layout.use": True,
    "axes.titlesize": 12,
    "axes.labelsize": 10,
    "xtick.labelsize": 9,
    "ytick.labelsize": 9,
    "legend.fontsize": 9,
    "legend.title_fontsize": 10,
    "grid.alpha": 1.0,
})

mpl.rcParams["axes.prop_cycle"] = cycler(color=["#000000", "#000000"])

# Mapas de color utilizados en el libro
_colormaps = {
    "black_and_blue": ["#000000", "#2f2fff"],
    "black_and_orange": ["#000000", "#D55E00"],
    "black": ["#000000", "#000000"],
    "black_and_2color": ["#000000", "#0072B2", "#D55E00"],
    "black_and_3color": ["#000000", "#D55E00", "#0072B2", "#009E73"],
    "black_and_4color": ["#000000", "#D55E00", "#0072B2", "#009E73", "#CC79A7"],
    "r_colors": ["#D55E00", "#0072B2", "#009E73", "#CC79A7"],
}

for name, colors in _colormaps.items():
    mpl.colormaps.register(
        mpl.colors.ListedColormap(colors, name=name),
        force=True,
    )

print("Configuración común cargada correctamente.")
```

## 5. Prueba de lectura con datos del Capítulo 2

El Capítulo 2 trabaja con distintos archivos CSV. Esta prueba usa `PBS_unparsed.csv`, uno de los datasets empleados para introducir manipulación de series de tiempo con `pandas`.

```python
pbs = (
    pd.read_csv(DATA_DIR / "PBS_unparsed.csv", parse_dates=["Month"])
    [["Month", "Concession", "Type", "ATC1", "ATC2", "Scripts", "Cost"]]
)

print("Dimensiones:", pbs.shape)
pbs.head()
```

## 6. Prueba gráfica

Esta gráfica sencilla verifica que `pandas`, `matplotlib` y la configuración visual del libro funcionan correctamente.

```python
a10 = pbs.loc[pbs["ATC2"] == "A10"].copy()

a10_monthly = (
    a10.groupby("Month", as_index=False)["Cost"]
       .sum()
)

ax = a10_monthly.plot(
    x="Month",
    y="Cost",
    legend=False,
    title="PBS — Costos mensuales de productos A10",
)
ax.set_xlabel("Mes")
ax.set_ylabel("Costo")
plt.show()
```
